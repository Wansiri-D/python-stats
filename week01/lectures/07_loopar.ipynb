```python
# enumerate() (การแจกแจงพรอมระบุตําแหน ง)

names = ["Joakim", "Hans", "Tommy"] 
for index, name in enumerate(names): 
    print(f"ตําแหน งที่ {index} คือ {name}")
```

```python
#  zip() (การรูดซิปจับคูขอมูล)

names = ["Joakim", "Hans"] 
scores = [85, 92] 

for name, score in zip(names, scores):
    print(f"{name} ไดคะแนน {score}")
```

```python
# การวนลูปใน Dictionary (ดวย .items)
student = {"name": "Bibbi", "age": 3} 

for key, value in student.items():
    print(f"{key}: {value}")
```

```python
# List Comprehension (การสรางลิสตแบบยอ)
# แทนที่จะเขียน for loop ปกติแลวใช .append()
# เราจะยอการหาเลขคูจาก 0-9 เหลือแคนี้: 

evens = [i for i in range(10) if i % 2 == 0]
```

```python
# for...else (ลูปที่มีเง ื ่อนไขจบ)
# บล็อก else จะทํางานก็ตอเมื ่อลูปนั ้นวิ่งจนจบตามปกติ โดยไมถูก break กลางทาง มักใชเวลาคนหา ขอมูล ถาหาไมเจอใหทําสิ่งน

for i in range(3):
    if i == 5: 
        break 
    else: 
        print("หาเลข 5 ไมเจอ ลูปวิ่ งจนจบเลย")
```
