```python
x = 10 # int (จํานวนเต็ม)
y = 3.14 # float (จํานวนทศนิยม)

print("x = ", type(x))
print("y = ", type(y))
```

```python
print(7 + 3)
print(7 - 3)
print(7 * 3)
print(7 / 3)
print(7 // 3) # heltalsdivision (การหารปัดเศษทิ้ง)
print(7 % 3) # modulus (หารเอาเศษ / rest)
print(7 ** 3) # upphöjt till (การยกกําลัง)
```

```python
x = 10
y = 3.5

print(float(x)) # แปลง int เป็น float
print(int(y)) # แปลง float เป็น int
```

```python
num = -7

print(abs(num)) # ค่าตัวเลขบวก (absolute value / ค่าสัมบูรณ์)
print(round(3.14159)) # # avrunda (ปัดเศษปกติ)
print(round(3.64159, 2)) # avrunda till 2 decimaler (ปัดเหลือ 2 ตําแหนง)
print(pow(2, 3)) # 2 upphöjt till 3 (ฟังกชันยกกําลัง)
```

```python
number = 10
number = number + 1 # แบบดั ้งเดิม (เอา 10 + 1 = 11 เก็บเขา number)
number += 1 # ทางลัด (มีความหมายเหมือนบรรทัดบนเปะ ตอนนี้เป็น 12)

number -= 1 # ลบออก 1 (เหลือ 11)
number -= 1 # ลบออกอีก 1 (เหลือ 10)

number //= 2 # หารปัดเศษดวย 2 (10 // 2 เหลือ 5)
number *= 3 # คูณดวย 3 (5 * 3 = 15)

print(number) # แสดงผลลัพธ์สุดท้าย (15)
```

```python
import math

print(math.sqrt(16)) # kvadratroten av 16 (รากที่สอง)
print(math.pi) # คาพายมาตรฐาน
print(math.floor(3.9)) # avrunda neråt (บังคับปัดเศษลง)
print(math.ceil(3.1)) # avrunda uppåt (บังคับปัดเศษขึ้น)
```
