```python
print(bool(0))       # เลขศูนย์
print(bool(-1))      # เลขติดลบ
print(bool(""))      # ข้อความว่างเปล่า
print(bool("Hej"))   # ข้อความที่มีตัวอักษร
print(bool([]))      # List (รายการ) ว่างๆ
print(bool([1, 2]))  # List ที่มีข้อมูล
print(bool(None))    # ค่าความว่างเปล่า (ไม่มีอะไรเลย)
```

I python räknas vissa värden automatiskt som falska t.ex:

0
En tom string ""
En tom lista []
None
Allt annat är sant (truthy)

Jämförelseoperationer

== Lika med
!= inte lika med
">" större än
"<" mindre än
">=" större eller lika
"<=" mindre eller lika

```python
ready = input("Are you ready to proceed? (y/n)")

if ready.lower() == "y":
    print("Great! here are the rest of the information.")
elif ready.lower() == "n":
    print("Ok, please read the info again.")
else:
    print("Oops, I don't understand what you mean.")
```

```python
age = 25
has_ticket = True

# ใช้ and (ต้องจริงทั้งคู่)
if age >= 18 and has_ticket:
    print("Welcome to the club")

# ใช้ or (จริงแค่อย่างใดอย่างหนึ่ง)
is_weekend = False
is_holiday = True
if is_weekend or is_holiday:
    print("No school today")

# ใช้ not (กลับจากเท็จเป็นจริง)
is_raining = False
if not is_raining:
    print("Let's go for a walk")
```
